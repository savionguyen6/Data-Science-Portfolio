Does the state's cost of living in the US relate to the poverty rate of different racial groups, and which states affect certain groups the most?

```python
import os
from dotenv import load_dotenv
import matplotlib.pyplot as plt
import pandas as pd
import requests
import plotly.express as px
from plotly.subplots import make_subplots
# Import all the necessary libraries and modules.
```

```python
load_dotenv()
API_KEY = os.getenv("CENSUS_API_KEY")
BEA_API_KEY = os.getenv("BEA_API_KEY")
# Looks at the .env file and retrieves the keys to access the API's.
```

```python
variables = [
    "NAME",          # State name
    "B17001A_001E",  # White Alone population
    "B17001A_002E",  # White Alone population below poverty
    "B17001B_001E",  # Black population
    "B17001B_002E",  # Black population below poverty
    "B17001D_001E",  # Asian population
    "B17001D_002E",  # Asian population below poverty
    "B17001I_001E",  # Hispanic/Latino population
    "B17001I_002E",  # Hispanic/Latino population below poverty
    "B19013_001E",   # Median household income
]
# All the necessary variables from the Census API.
```

```python
census_url = (f"https://api.census.gov/data/2023/acs/acs1"f"?get={','.join(variables)}&for=state:*&key={API_KEY}")
bea_url = ("https://apps.bea.gov/api/data/"f"?UserID={BEA_API_KEY}""&method=GETDATA""&datasetname=Regional""&TableName=SARPP""&LineCode=1""&GeoFips=STATE""&Year=2023""&ResultFormat=JSON")
# The actual information or location to access to the API's.
```

```python
census_response = requests.get(census_url)
census_data = census_response.json()
bea_response = requests.get(bea_url)
bea_data = bea_response.json()
# Gets the API's and converts to python data.
```

```python
census_data
# all the information from the variables that are from this census.
```

```python
bea_data 
# # Viewing how to retrieve the data in this BEA API (Different from Census API).
```

```python
df = pd.DataFrame(census_data[1:], columns=census_data[0])
df.head()
# Simple retrieve from the first column and then on. 
# Shows the first five top rows.
```

```python
bea_results = bea_data["BEAAPI"]["Results"]["Data"]
bea_df = pd.DataFrame(bea_results)
bea_df.head()
# Digging through the layers to get to the data.
```

```python
bea_df = bea_df[["GeoName", "DataValue"]] # Getting specific columns that will be used in the research analysis.
```

```python
bea_df = bea_df.rename(columns = {"GeoName": "NAME", "DataValue": "RPP"}) 
# Need to change the "GeoName" to match "NAME", to allow a merge later with the other table (Census API). 
# Also Chaning DataValue to RPP (Regional Price Parities), the mesure of cost of living for this analysis. Purchasing power by understanding how much the price of goods and services change.
```

```python
bea_df.head() # Shows the first five rows.
```

```python
df = df.merge(bea_df, on = "NAME", how = "left") # Merges the two tables togther based on the same column "NAME"
```

```python
df[["NAME", "RPP"]].head(10) # Shows the first 10 rows of the information after merging or joining the two tables.
```

```python
num_com = df.columns[1:]
df[num_com] = df[num_com].apply(pd.to_numeric, errors = "coerce")
df = df.rename(columns = {
    "B17001A_002E": "White_poverty_population",
    "B17001B_002E": "Black_poverty_population",
    "B17001D_002E": "Asian_poverty_population",
    "B17001I_002E": "Hispanic_poverty_population",
    "B19013_001E": "Median Household Income"
})
# Converts data in the table that are string but look like numbers into int or float and if there is a null or N/A answer NAN.
# Changes variable names into readable names
```

```python
df["White Poverty Rate"] = (df["White_poverty_population"] / df["B17001A_001E"] * 100).round(1)
df["Black Poverty Rate"] = (df["Black_poverty_population"] / df["B17001B_001E"] * 100).round(1)
df["Asian Poverty Rate"] = (df["Asian_poverty_population"] / df["B17001D_001E"] * 100).round(1)
df["Hispanic Poverty Rate"] = (df["Hispanic_poverty_population"] / df["B17001I_001E"] * 100).round(1)
# Finds the precentage rate of poverty for each racial group by dividing the total amount below poverty by the total amount.
# As well round to the tenth decimal number.
```

```python
state = {
    1: "AL",
    2: "AK",
    4: "AZ",
    5: "AR",
    6: "CA",
    8: "CO",
    9: "CT",
    10: "DE",
    11: "DC",
    12: "FL",
    13: "GA",
    15: "HI",
    16: "ID",
    17: "IL",
    18: "IN",
    19: "IA",
    20: "KS",
    21: "KY",
    22: "LA",
    23: "ME",
    24: "MD",
    25: "MA",
    26: "MI",
    27: "MN",
    28: "MS",
    29: "MO",
    30: "MT",
    31: "NE",
    32: "NV",
    33: "NH",
    34: "NJ",
    35: "NM",
    36: "NY",
    37: "NC",
    38: "ND",
    39: "OH",
    40: "OK",
    41: "OR",
    42: "PA",
    44: "RI",
    45: "SC",
    46: "SD",
    47: "TN",
    48: "TX",
    49: "UT",
    50: "VT",
    51: "VA",
    53: "WA",
    54: "WV",
    55: "WI",
    56: "WY"
}

df["State"] = df["state"].map(state)
```

```python
figRPP = px.choropleth(df, locations = "State", locationmode = "USA-states", color_continuous_scale = "Blues", color = "RPP", scope = "usa", title = "The RPP rate for each State", height = 900, width = 1700)
figRPP.show()

fig1 = px.choropleth(df, locations = "State", locationmode = "USA-states", color = "Black Poverty Rate")

fig2 = px.choropleth(df, locations = "State", locationmode = "USA-states", color = "White Poverty Rate")

fig3 = px.choropleth(df,locations = "State", locationmode = "USA-states", color = "Asian Poverty Rate")

fig4 = px.choropleth(df, locations = "State", locationmode = "USA-states", color = "Hispanic Poverty Rate")

figFRS = make_subplots(rows = 2, cols = 2,
    specs = [[{"type": "choropleth"}, {"type": "choropleth"}], [{"type": "choropleth"}, {"type": "choropleth"}]],
    subplot_titles = ["Black Poverty Rate by State", "White Poverty Rate by State", "Asian Poverty Rate by State", "Hispanic Poverty Rate by State"])
# Poverty Rate graphs
figFRS.add_trace(fig1.data[0], row = 1, col = 1)
figFRS.add_trace(fig2.data[0], row = 1, col = 2)
figFRS.add_trace(fig3.data[0], row = 2, col = 1)
figFRS.add_trace(fig4.data[0], row = 2, col = 2)

figFRS.update_geos(scope = "usa")
figFRS.update_layout(height = 900, width = 1700)

figFRS.show()
# Creates an interactive graphs with hovering text as well shows the distribution/spread of poverty rates for each racial group for each state.
# Creates an interactive graphs with hovering text as well shows the RPP for each state.
```
